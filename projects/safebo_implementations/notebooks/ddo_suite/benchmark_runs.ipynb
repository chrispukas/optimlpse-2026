```python
from typing import Callable

from ddo_suite.problems import (
    Sphere, 
    ExothermicCstr,
    PidTuning,
    )
from ddo_suite.algorithms import (
    ALGORITHM_IDS,
    __all__,

    random_search,
    safeopt,
    goose,
    )
from ddo_suite.experiments import (
    run_benchmark, 
    aggregate_results, 
    ExperimentCache
    )
from ddo_suite.experiments.plotting import (
    save_figure
)
```

```python
problems: list[type] = \
    [
        Sphere,
        ExothermicCstr,
        PidTuning,
    ]
algorithms: list[Callable] = \
    [
        random_search,
        safeopt,
        goose,
    ]

cache: ExperimentCache = ExperimentCache("cache/")
results = run_benchmark(
    problems=problems,
    algorithms=algorithms,
    n_x=2, seeds=range(10), max_evals=(200), cache=cache, verbose=True
)
agg = aggregate_results(results=results)
print(agg.ranking("ackley", 5))
```
